# Smoke Test: PSD and FOOOF

Run PSD through the stable workflow and FOOOF through the legacy helper on a small epoch subset.

```python
from pathlib import Path
from LFPAnalysis import LoadConfig, SpectralConfig, compute_spectral_features, load_lfp
from LFPAnalysis import analysis_utils

epochs = load_lfp(LoadConfig(path=Path('../../data/sample_feedback_start-epo.fif'), file_format='mne'))
epochs = epochs.copy().pick(epochs.ch_names[:4])[:10]
psd = compute_spectral_features(epochs, SpectralConfig(enabled=True, method='psd', fmin=2, fmax=40))
print(psd['method'])
fooof_group, fooof_table = analysis_utils.FOOOF_compute_epochs(
    epochs,
    tmin=float(epochs.times[0]),
    tmax=float(epochs.times[-1]),
    peak_width_limits=(1, 12),
    min_peak_height=0.0,
    peak_threshold=2.0,
    max_n_peaks=4,
    freq_range=(2, 40),
)
print(fooof_table.head())
```
