# Smoke Test: PSD and FOOOF

```python
from pathlib import Path
from LFPAnalysis import build_analysis_config, load_lfp, run_analysis
from LFPAnalysis.config import LoadConfig

epochs = load_lfp(
    LoadConfig(path=Path('../../data/sample_feedback_start-epo.fif'), file_format='mne', preload=True)
)
epochs = epochs[:5].pick(['racas1-racas2'])
psd_result = run_analysis(epochs, build_analysis_config(spectral_method='psd', fmin=1.0, fmax=40.0))
print('psd', psd_result.spectral['spectrum'].get_data().shape)
fooof_result = run_analysis(epochs, build_analysis_config(spectral_method='fooof', fooof_range=(1.0, 40.0)))
print('fooof rows', len(fooof_result.spectral['table']))
```
