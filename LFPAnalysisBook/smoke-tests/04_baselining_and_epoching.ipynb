# Smoke Test: Baselining and Epoching

Create synthetic task epochs from a continuous sample recording and apply a z-score baseline.

```python
from pathlib import Path
from LFPAnalysis import BaselineConfig, EpochConfig, LoadConfig, baseline_lfp, load_lfp, make_epochs

raw = load_lfp(LoadConfig(path=Path('../../data/sample_ieeg_continuous_rest.fif'), file_format='mne'))
raw.crop(tmax=min(20.0, raw.times[-1]))
epochs = make_epochs(raw, EpochConfig(enabled=True, event_name='demo', event_times=[5.0, 10.0, 15.0], tmin=-0.5, tmax=1.0))
baselined, summary = baseline_lfp(epochs, BaselineConfig(mode='zscore', enabled=True, baseline_window=(-0.5, 0.0)))
print(baselined)
print(summary.head())
```
