# Smoke Test: Baselining and Epoching

```python
import pandas as pd
from pathlib import Path
from LFPAnalysis import load_lfp, make_epochs, baseline_lfp
from LFPAnalysis.config import LoadConfig, EpochConfig, BaselineConfig

beh = pd.read_csv(Path('../../data/sample_beh.csv'))
raw = load_lfp(LoadConfig(path=Path('../../data/sample_ieeg_bp.fif'), file_format='mne'))
raw.crop(tmax=300)
epochs = make_epochs(
    raw,
    EpochConfig(
        enabled=True,
        event_name='feedback_start',
        event_times=beh['feedback_start'].tolist()[:3],
        tmin=-0.5,
        tmax=1.0,
    ),
)
ep_base, summary = baseline_lfp(epochs, BaselineConfig(mode='zscore', enabled=True, baseline_window=(-0.5, 0.0)))
print(ep_base, summary.head())
```
