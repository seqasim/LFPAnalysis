# Smoke Test: Connectivity and Surrogates

Use a lightweight synthetic example to validate surrogate generation and connectivity dependencies.

```python
import numpy as np
import mne
from mne_connectivity import spectral_connectivity_epochs
from LFPAnalysis.oscillation_utils import make_surrogate_arrays

rng = np.random.default_rng(7)
sfreq = 200.0
times = np.arange(0, 2, 1 / sfreq)
base = np.sin(2 * np.pi * 10 * times)
data = []
for _ in range(8):
    x = base + 0.05 * rng.standard_normal(len(times))
    y = np.roll(x, 3) + 0.05 * rng.standard_normal(len(times))
    data.append(np.vstack([x, y]))
data = np.asarray(data)
info = mne.create_info(['seed', 'target'], sfreq, ch_types=['seeg', 'seeg'])
epochs = mne.EpochsArray(data, info)
conn = spectral_connectivity_epochs(epochs, method='coh', sfreq=sfreq, fmin=8, fmax=12, verbose=False)
surr = list(make_surrogate_arrays(data[:, 0, :], method='swap_epochs', n_shuffles=3, return_generator=False))
print(conn.get_data().shape)
print(len(surr), surr[0].shape)
```
