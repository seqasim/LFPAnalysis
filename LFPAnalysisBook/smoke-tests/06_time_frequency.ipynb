# Smoke Test: Time-Frequency Analysis

Compute a bounded Morlet transform on a small epoch subset.

```python
from pathlib import Path
import numpy as np
from LFPAnalysis import LoadConfig, load_lfp

epochs = load_lfp(LoadConfig(path=Path('../../data/sample_feedback_start-epo.fif'), file_format='mne'))
epochs = epochs.copy().pick(epochs.ch_names[:2])[:5]
tfr = epochs.compute_tfr(method='morlet', freqs=np.arange(4, 31, 4), n_cycles=3, average=True)
print(tfr)
print(tfr.data.shape)
```
