# Smoke Test: Reference and Artifacts

```python
from pathlib import Path
from LFPAnalysis import load_lfp, preprocess_lfp, detect_artifacts
from LFPAnalysis.config import LoadConfig, ReferenceConfig, ArtifactConfig

raw = load_lfp(LoadConfig(path=Path('../../data/sample_ieeg_bp.fif'), file_format='mne'))
raw.crop(tmax=60)
ref = preprocess_lfp(raw, ReferenceConfig(method='none'))
tables = detect_artifacts(ref, ArtifactConfig(methods=['misc'], misc_peak_thresh=8.0))
print(tables['misc'].head(), len(tables['misc']))
```
