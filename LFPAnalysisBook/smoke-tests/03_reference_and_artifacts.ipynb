# Smoke Test: Reference and Artifact Detection

Use the workflow API with a no-op reference and the built-in miscellaneous artifact detector.

```python
from pathlib import Path
from LFPAnalysis import ArtifactConfig, LoadConfig, ReferenceConfig, detect_artifacts, load_lfp, preprocess_lfp

raw = load_lfp(LoadConfig(path=Path('../../data/sample_ieeg_continuous_rest.fif'), file_format='mne'))
raw.crop(tmax=min(10.0, raw.times[-1]))
referenced = preprocess_lfp(raw, ReferenceConfig(method='none'))
artifacts = detect_artifacts(referenced, ArtifactConfig(methods=['misc'], misc_peak_thresh=8.0))
misc_table = artifacts['misc']
print(misc_table.head())
print('misc rows:', len(misc_table))
```
