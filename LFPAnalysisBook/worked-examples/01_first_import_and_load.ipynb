# Worked Example: First Import and First Load

## Goal
Load bundled sample data through the beginner-facing API and inspect the returned object.

## Expected input files
- `../../data/sample_ieeg.fif`

## What this notebook teaches
- how to start with a builder instead of manual configs
- how to inspect the result object
- how to tell whether loading succeeded for the right reasons


```python
from pathlib import Path
from LFPAnalysis import build_basic_pipeline_config, run_pipeline

config = build_basic_pipeline_config(Path('../../data/sample_ieeg.fif'), file_format='mne')
result = run_pipeline(config)
result.raw
```

## Inspect the result

The first thing to inspect is not a plot. It is the object contract: sampling rate, channel names, and the metadata recorded by the pipeline.

```python
print(result.raw.info['sfreq'])
print(result.raw.ch_names[:5])
print(result.metadata)
```

## Meaningful parameter change

Now change one thing that matters but is still safe: pick a subset of channels before loading.

```python
from LFPAnalysis import LoadConfig, PipelineConfig, ReferenceConfig, ArtifactConfig, BaselineConfig, EpochConfig, SpectralConfig, run_pipeline

config_subset = PipelineConfig(
    load=LoadConfig(path=Path('../../data/sample_ieeg.fif'), file_format='mne', pick_channels=result.raw.ch_names[:3]),
    reference=ReferenceConfig(method='none'),
    artifact=ArtifactConfig(methods=['none']),
    baseline=BaselineConfig(mode='none', enabled=False),
    epoch=EpochConfig(enabled=False),
    spectral=SpectralConfig(enabled=False, method='none'),
)
subset = run_pipeline(config_subset)
print(subset.raw.ch_names)
```

## Failure interpretation

If this notebook fails, the most likely problems are a missing dependency or a wrong path. Do not move on to referencing or epoching until load works cleanly.
