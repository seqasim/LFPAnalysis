# Worked Example: Photodiode Synchronization

## Goal
Align behavioral sync pulses to the neural photodiode recording.


```python
import pandas as pd
import mne
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from LFPAnalysis import sync_utils

beh_ts = pd.read_csv(Path('../../data/sample_ts.csv'))['beh_ts'].values
photodiode = mne.io.read_raw_fif(Path('../../data/sample_photodiode.fif'), preload=True, verbose=False)
slope, offset = sync_utils.synchronize_data(beh_ts=beh_ts, mne_sync=photodiode, sync_source='photodiode')
print(f'slope={slope:.4f}, offset={offset:.2f}')
```

## Plot first matched pulses

```python
neural_ts = sync_utils.get_neural_ts_photodiode(photodiode)
n_show = min(30, len(beh_ts), len(neural_ts))
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(np.arange(n_show), beh_ts[:n_show], 'o-', label='behavioral')
ax.plot(np.arange(n_show), neural_ts[:n_show], 's-', label='neural (photodiode)')
ax.set(xlabel='Pulse index', ylabel='Time (s)', title='Sync pulse alignment')
ax.legend()
fig.tight_layout()
plt.show()
```
