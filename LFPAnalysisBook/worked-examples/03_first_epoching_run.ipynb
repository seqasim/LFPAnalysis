# Worked Example: First Event-Locked Run

## Goal
Create epochs from a continuous recording using the beginner-friendly event-locked builder.

## Expected input files
- `../../data/sample_ieeg_continuous_rest.fif`

```python
from pathlib import Path
from LFPAnalysis import build_event_locked_pipeline_config, run_pipeline

config = build_event_locked_pipeline_config(
    Path('../../data/sample_ieeg_continuous_rest.fif'),
    file_format='mne',
    event_name='demo',
    event_times=[5.0, 10.0, 15.0],
    baseline_mode='zscore',
    baseline_window=(-0.5, 0.0),
)
result = run_pipeline(config)
result.epochs
```

## Inspect the result

Check the number of epochs, the time window, and the baseline summary before moving to any spectral analysis.

```python
print(len(result.epochs))
print(result.epochs.times[[0, -1]])
print(result.baseline_summary.head())
```

## Meaningful parameter change

Shorten the epoch window and compare the time axis.

```python
config_short = build_event_locked_pipeline_config(
    Path('../../data/sample_ieeg_continuous_rest.fif'),
    file_format='mne',
    event_name='demo',
    event_times=[5.0, 10.0, 15.0],
    tmin=-0.25,
    tmax=0.75,
    baseline_mode='zscore',
    baseline_window=(-0.25, 0.0),
)
short_result = run_pipeline(config_short)
print(short_result.epochs.times[[0, -1]])
```

## Failure interpretation

If epoching fails, check whether your event timestamps are in seconds and whether the requested window fits within the recording.
