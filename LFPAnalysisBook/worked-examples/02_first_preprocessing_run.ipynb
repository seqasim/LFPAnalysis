# Worked Example: Bipolar Reference and Artifact QC

## Goal
Apply bipolar referencing to the real recording and run misc-artifact detection.


```python
from pathlib import Path
from LFPAnalysis import build_basic_pipeline_config, run_pipeline

config = build_basic_pipeline_config(
    Path('../../data/sample_ieeg.fif'),
    file_format='mne',
    reference_method='bipolar',
    electrode_path=Path('../../data/sample_labels.xlsx'),
    artifact_methods=['misc'],
)
result = run_pipeline(config)
print(f'Monopolar {len(result.raw.ch_names)} ch → Bipolar {len(result.referenced.ch_names)} ch')
artifact_table = result.artifact_tables['misc']
print(f'Misc artifacts: {len(artifact_table)} events')
print(artifact_table.head())
```

## Plot artifact density by channel

```python
import matplotlib.pyplot as plt

if len(artifact_table):
    counts = artifact_table.groupby('channel').size().sort_values(ascending=False)
    counts.head(8).plot(kind='bar', figsize=(8, 3), title='Misc artifacts per channel')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
else:
    print('No misc artifacts detected at default threshold.')
```
