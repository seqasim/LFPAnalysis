# Worked Example: Load the Gambling-Task Recording

## Goal
Load the real 22-channel `sample_ieeg.fif` and inspect sampling rate, channels, and duration.

## Expected input files
- `../../data/sample_ieeg.fif`


```python
from pathlib import Path
from LFPAnalysis import build_basic_pipeline_config, run_pipeline

config = build_basic_pipeline_config(Path('../../data/sample_ieeg.fif'), file_format='mne')
result = run_pipeline(config)
raw = result.raw
print(f'sfreq={raw.info["sfreq"]} Hz, n_ch={len(raw.ch_names)}, duration={raw.n_times/raw.info["sfreq"]:.1f} s')
print('Channels:', raw.ch_names[:8], '…')
print('preload in metadata:', result.metadata.get('preload'))
```

## Plot a raw trace and sanity-check PSD

```python
import matplotlib.pyplot as plt
import numpy as np

sfreq = raw.info['sfreq']
start = int(240 * sfreq)
stop = start + int(2 * sfreq)
data = raw.get_data(picks=['racas1'])[0, start:stop]
times = np.arange(len(data)) / sfreq

fig, axes = plt.subplots(1, 2, figsize=(10, 3))
axes[0].plot(times, data, 'k', lw=0.8)
axes[0].set(xlabel='Time (s)', ylabel='Amplitude', title='racas1 around first feedback')
psd = raw.compute_psd(fmin=1, fmax=80, picks=['racas1'])
axes[1].semilogy(psd.freqs, psd.get_data()[0])
axes[1].set(xlabel='Frequency (Hz)', ylabel='PSD', title='Welch PSD')
fig.tight_layout()
plt.show()
```
